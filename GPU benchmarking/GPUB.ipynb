```python
import urllib.request
import os

url = "https://www.cs.toronto.edu/~kriz/cifar-10-python.tar.gz"
target_dir = os.path.expanduser(r"~\.keras\datasets")
os.makedirs(target_dir, exist_ok=True)
target_file = os.path.join(target_dir, "cifar-10-python.tar.gz")

print("Downloading CIFAR-10 cleanly via Python...")
urllib.request.urlretrieve(url, target_file)
print("Download complete! File saved to:", target_file)
```
